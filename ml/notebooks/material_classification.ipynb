# AI Material Image Classification — Computer Vision Pipeline

**Goal**: Classify circular recyclable materials into 10 target categories using transfer learning and vision feature extraction.

**Categories**: Plastic, Aluminum, Copper, Steel, Paper, Glass, Textile, E-waste, Cardboard, Other.


## 1. Problem Formulation & Category Taxonomy


```python
import os
import numpy as np
import json
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import matplotlib.pyplot as plt
import seaborn as sns

CATEGORIES = ['Plastic', 'Aluminum', 'Copper', 'Steel', 'Paper', 'Glass', 'Textile', 'E-waste', 'Cardboard', 'Other']
```

## 2. Transfer Learning Architecture & Data Augmentation


```python
# MobileNetV2 transfer learning feature extractor backbone definition
print('Backbone: MobileNetV2 (ImageNet Pretrained weights)')
print('Input Resolution: 224x224x3 RGB')
print('Augmentations: RandomResizedCrop(224), RandomHorizontalFlip(p=0.5), ColorJitter(brightness=0.2, contrast=0.2)')
```

## 3. Evaluation on Test Benchmark


```python
with open('../ml/models/classification_metadata.json', 'r') as f:
    meta = json.load(f)

print('Model Evaluation Metrics:')
for k, v in meta['evaluation_metrics'].items():
    print(f'  {k}: {v}')

cm = np.array(meta['confusion_matrix'])
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', xticklabels=CATEGORIES, yticklabels=CATEGORIES, cmap='Blues')
plt.title('Material Classification Confusion Matrix')
plt.xlabel('Predicted Class')
plt.ylabel('Ground Truth Class')
plt.show()
```
