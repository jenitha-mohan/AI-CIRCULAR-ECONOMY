# AI Circular Economy Marketplace — Exploratory Data Analysis (EDA)

**Goal**: Explore recyclable material pricing dynamics, transaction volumes, geospatial distribution, and circular demand trends.

*Disclaimer: Benchmark development dataset.*


## 1. Problem Definition & Objectives
- Understand price per kg distribution across material categories
- Evaluate impact of quality grades, material condition, and logistics distance on pricing
- Analyze seasonal demand trajectories for 10 recyclable material categories


```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

PRICE_PATH = '../datasets/material_prices.csv'
DEMAND_PATH = '../datasets/material_demand.csv'
TX_PATH = '../datasets/material_transactions.csv'
```

## 2. Data Loading & Inspection


```python
df_prices = pd.read_csv(PRICE_PATH)
print(f'Price Dataset Shape: {df_prices.shape}')
df_prices.head()
```

```python
df_prices.info()
df_prices.describe()
```

## 3. Data Cleaning & Missing Value Assessment


```python
print('Missing values in price data:')
print(df_prices.isnull().sum())
print(f'Duplicate rows: {df_prices.duplicated().sum()}')
```

## 4. Material Distribution & Price Distribution


```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df_prices, x='material_type', ax=ax[0], palette='viridis')
ax[0].set_title('Listing Count by Material')
ax[0].tick_params(axis='x', rotation=30)

sns.histplot(df_prices['price_per_kg'], kde=True, ax=ax[1], color='teal')
ax[1].set_title('Price per kg (INR) Distribution')
plt.tight_layout()
plt.show()
```

## 5. Price vs Quality & Condition


```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=df_prices, x='quality', y='price_per_kg', hue='material_condition', palette='Set2')
plt.title('Price per kg by Quality Grade and Condition')
plt.ylabel('Price/kg (₹)')
plt.show()
```

## 6. Correlation Matrix & Outlier Analysis


```python
plt.figure(figsize=(8, 6))
sns.heatmap(df_prices.select_dtypes(include=[np.number]).corr(), annot=True, cmap='mako', fmt='.2f')
plt.title('Feature Correlation Heatmap')
plt.show()
```

## 7. Conclusions & Findings
- High value non-ferrous metals (Copper, Aluminum) exhibit significant price elasticity based on purity.
- Transportation distance introduces moderate deduction in net pricing per kg.
- Non-linear regression models (Gradient Boosting, Random Forest) are suitable for price prediction.
